```python
train = train[~((train["GrLivArea"] > 4000) & (train["SalePrice"] < 300000))].reset_index(drop=True)
print(train.shape)   # (1458, 81)
```

```python
# Save the target and separate it from the features
y = np.log1p(train["SalePrice"])
X = train.drop(columns=["SalePrice", "Id"])
test_ids = test["Id"]
X_test = test.drop(columns=["Id"])

none_cols = ["PoolQC", "MiscFeature", "Alley", "Fence", "FireplaceQu", "MasVnrType",
             "GarageType", "GarageFinish", "GarageQual", "GarageCond",
             "BsmtQual", "BsmtCond", "BsmtExposure", "BsmtFinType1", "BsmtFinType2"]
zero_cols = ["MasVnrArea", "GarageYrBlt", "GarageCars", "GarageArea",
             "BsmtFinSF1", "BsmtFinSF2", "BsmtUnfSF", "TotalBsmtSF",
             "BsmtFullBath", "BsmtHalfBath"]

# Statistics learned on TRAIN only
lot_median = X.groupby("Neighborhood")["LotFrontage"].median()
global_lot = X["LotFrontage"].median()

def fill_missing(df):
    df = df.copy()
    df[none_cols] = df[none_cols].fillna("None")
    df[zero_cols] = df[zero_cols].fillna(0)
    df["LotFrontage"] = df["LotFrontage"].fillna(df["Neighborhood"].map(lot_median)).fillna(global_lot)
    return df

X = fill_missing(X)
X_test = fill_missing(X_test)

# Remaining gaps: mode for text columns, median for numbers (values from train)
for col in X.columns:
    if X[col].isna().any() or X_test[col].isna().any():
        fill = X[col].mode()[0] if X[col].dtype == "object" or str(X[col].dtype) == "str" else X[col].median()
        X[col] = X[col].fillna(fill)
        X_test[col] = X_test[col].fillna(fill)

print(X.isna().sum().sum(), X_test.isna().sum().sum())   # both should be 0
```

```python
# 0. Stack train and test so they get identical columns after encoding.
#    The target y is NOT involved, so no label leakage.
n_train = len(X)
all_data = pd.concat([X, X_test], axis=0, ignore_index=True)

# 1. MSSubClass is a code (20, 60, 120...), not a quantity -> category
all_data["MSSubClass"] = all_data["MSSubClass"].astype(str)

# 2. Quality scales -> ordered numbers (higher = better)
qual_map = {"None": 0, "Po": 1, "Fa": 2, "TA": 3, "Gd": 4, "Ex": 5}
qual_cols = ["ExterQual", "ExterCond", "BsmtQual", "BsmtCond", "HeatingQC",
             "KitchenQual", "FireplaceQu", "GarageQual", "GarageCond", "PoolQC"]
for col in qual_cols:
    all_data[col] = all_data[col].map(qual_map)

exposure_map = {"None": 0, "No": 1, "Mn": 2, "Av": 3, "Gd": 4}
all_data["BsmtExposure"] = all_data["BsmtExposure"].map(exposure_map)

# 3. New features
all_data["TotalSF"] = all_data["TotalBsmtSF"] + all_data["1stFlrSF"] + all_data["2ndFlrSF"]
all_data["TotalBathrooms"] = (all_data["FullBath"] + 0.5 * all_data["HalfBath"]
                              + all_data["BsmtFullBath"] + 0.5 * all_data["BsmtHalfBath"])
all_data["HouseAge"] = (all_data["YrSold"] - all_data["YearBuilt"]).clip(lower=0)
all_data["YearsSinceRemodel"] = (all_data["YrSold"] - all_data["YearRemodAdd"]).clip(lower=0)
all_data["HasPool"] = (all_data["PoolArea"] > 0).astype(int)
all_data["HasGarage"] = (all_data["GarageArea"] > 0).astype(int)
all_data["HasFireplace"] = (all_data["Fireplaces"] > 0).astype(int)

# GarageYrBlt was filled with 0 for "no garage", which is misleading; HasGarage replaces it
all_data = all_data.drop(columns=["GarageYrBlt"])

# 4. Log-transform skewed numeric columns (skewness measured on TRAIN rows only)
flags = ["HasPool", "HasGarage", "HasFireplace"]
num_cols = [c for c in all_data.select_dtypes(include="number").columns if c not in flags]
skew = all_data.loc[:n_train - 1, num_cols].skew()
skewed = skew[skew.abs() > 0.75].index
all_data[skewed] = np.log1p(all_data[skewed])
print(len(skewed), "columns log-transformed")

# 5. One-hot encode the remaining text columns
cat_cols = all_data.select_dtypes(exclude="number").columns
all_data = pd.get_dummies(all_data, columns=cat_cols, dtype=int)

# 6. Split back into train and test
X_enc = all_data.iloc[:n_train].copy()
X_test_enc = all_data.iloc[n_train:].copy()

print(X_enc.shape, X_test_enc.shape)
print("Same columns:", list(X_enc.columns) == list(X_test_enc.columns))
print("NaN:", X_enc.isna().sum().sum(), X_test_enc.isna().sum().sum())
```

```python
from sklearn.model_selection import KFold, cross_val_score
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import RobustScaler
from sklearn.dummy import DummyRegressor
from sklearn.linear_model import LinearRegression, Ridge, Lasso, ElasticNet
from sklearn.ensemble import RandomForestRegressor, GradientBoostingRegressor
from xgboost import XGBRegressor
from lightgbm import LGBMRegressor
kf = KFold(n_splits=5, shuffle=True, random_state=42)
def cv_rmse(model):
    scores = cross_val_score(model, X_enc, y, scoring="neg_root_mean_squared_error", cv=kf, n_jobs=-1)
    return -scores   
```

```python
models = {
    "Baseline (mean)":   DummyRegressor(strategy="mean"),
    "Linear Regression": make_pipeline(RobustScaler(), LinearRegression()),
    "Ridge":             make_pipeline(RobustScaler(), Ridge(alpha=10)),
    "Lasso":             make_pipeline(RobustScaler(), Lasso(alpha=0.0005, max_iter=20000)),
    "ElasticNet":        make_pipeline(RobustScaler(), ElasticNet(alpha=0.0005, l1_ratio=0.9, max_iter=20000)),
    "Random Forest":     RandomForestRegressor(n_estimators=300, random_state=42, n_jobs=-1),
    "Gradient Boosting": GradientBoostingRegressor(n_estimators=500, learning_rate=0.05, max_depth=3, subsample=0.8, random_state=42),
    "XGBoost":           XGBRegressor(n_estimators=1000, learning_rate=0.03, max_depth=3, subsample=0.8, colsample_bytree=0.5, random_state=42),
    "LightGBM":          LGBMRegressor(n_estimators=1000, learning_rate=0.03, num_leaves=8, subsample=0.8, colsample_bytree=0.5, random_state=42, verbose=-1),
}
```

```python
results = []
for name, model in models.items():
    s = cv_rmse(model)
    results.append({"Model": name, "CV RMSE (mean)": s.mean(), "Std": s.std()})
    print(f"{name:20s} {s.mean():.4f} (+/- {s.std():.4f})")

results_df = pd.DataFrame(results).sort_values("CV RMSE (mean)").reset_index(drop=True)
display(results_df)
```

```python
plt.figure(figsize=(8, 5))
sns.barplot(x="CV RMSE (mean)", y="Model", data=results_df, color="steelblue")
plt.title("5-fold CV RMSE on log price (lower is better)")
plt.tight_layout()
plt.show()
```

```python
from sklearn.model_selection import GridSearchCV
from sklearn.ensemble import VotingRegressor

def tune(model, grid):
    gs = GridSearchCV(model, grid, scoring="neg_root_mean_squared_error", cv=kf, n_jobs=-1)
    gs.fit(X_enc, y)
    print(gs.best_params_, round(-gs.best_score_, 4))
    return gs.best_estimator_

lasso = tune(make_pipeline(RobustScaler(), Lasso(max_iter=20000)),
             {"lasso__alpha": [0.0002, 0.0005, 0.001, 0.002]})
ridge = tune(make_pipeline(RobustScaler(), Ridge()),
             {"ridge__alpha": [5, 10, 20, 30]})
lgbm = tune(LGBMRegressor(n_estimators=1000, colsample_bytree=0.5, random_state=42, verbose=-1),
            {"learning_rate": [0.02, 0.03, 0.05], "num_leaves": [4, 8, 16]})

# Ensemble = simple average of the three tuned models
ensemble = VotingRegressor([("lasso", lasso), ("ridge", ridge), ("lgbm", lgbm)])
s = cv_rmse(ensemble)
print(f"Ensemble CV RMSE: {s.mean():.4f} (+/- {s.std():.4f})")
```

```python
from sklearn.model_selection import cross_val_predict
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score

# Each house is predicted by a model that never saw it
oof = cross_val_predict(ensemble, X_enc, y, cv=kf, n_jobs=-1)

actual = np.expm1(y)       # back to dollars
pred = np.expm1(oof)

print("RMSE (log price):", round(np.sqrt(mean_squared_error(y, oof)), 4))
print("R2 (dollars):    ", round(r2_score(actual, pred), 4))
print("MAE (dollars):   ", round(mean_absolute_error(actual, pred)))
```

```python
resid = y - oof
fig, ax = plt.subplots(1, 2, figsize=(13, 5))

ax[0].scatter(actual, pred, alpha=0.5)
lims = [actual.min(), actual.max()]
ax[0].plot(lims, lims, color="red")
ax[0].set_xlabel("Actual price"); ax[0].set_ylabel("Predicted price")
ax[0].set_title("Predicted vs actual")

ax[1].scatter(oof, resid, alpha=0.5)
ax[1].axhline(0, color="red")
ax[1].set_xlabel("Predicted log price"); ax[1].set_ylabel("Residual (log)")
ax[1].set_title("Residuals")
plt.tight_layout()
plt.show()
```

```python
final_lgbm = lgbm.fit(X_enc, y)
imp = pd.Series(final_lgbm.feature_importances_, index=X_enc.columns).sort_values(ascending=False).head(15)

plt.figure(figsize=(8, 6))
sns.barplot(x=imp.values, y=imp.index, color="steelblue")
plt.title("Top 15 features (LightGBM)")
plt.tight_layout()
plt.show()
```
